```python
def convert_to_datetime(df: pd.DataFrame, columns: list[str]) -> pd.DataFrame:
    """
    Convert specified columns to datetime.

    Parameters:
        df (pd.DataFrame): Input DataFrame.
        columns (list[str]): Columns to convert.

    Returns:
        pd.DataFrame: DataFrame with converted columns.
    """
    for col in columns:
        df[col] = pd.to_datetime(df[col])

    return df

def convert_to_nullable_integer(
    df: pd.DataFrame,
    columns: list[str]
) -> pd.DataFrame:
    """
    Convert specified columns to nullable integer type.

    Parameters:
        df (pd.DataFrame): Input DataFrame.
        columns (list[str]): Columns to convert.

    Returns:
        pd.DataFrame: DataFrame with converted columns.
    """
    for col in columns:
        df[col] = df[col].astype("Int64")

    return df

def convert_to_category(
    df: pd.DataFrame,
    columns: list[str]
) -> pd.DataFrame:
    """
    Convert specified columns to categorical type.

    Parameters:
        df (pd.DataFrame): Input DataFrame.
        columns (list[str]): Columns to convert.

    Returns:
        pd.DataFrame: DataFrame with converted columns.
    """
    for col in columns:
        df[col] = df[col].astype("category")

    return df
```

## Time Consistency Check

Since the `start_time`, `end_time`, and `duration_sec` columns are related, we can verify whether the end time is consistent with the start time and trip duration.

Because the time values represent time within an hour, the calculation is performed using a circular 3600-second format.

```python
def time_to_seconds(
    time_series: pd.Series
) -> pd.Series:
    """
    Convert MM:SS.f time values to seconds.

    Parameters:
        time_series (pd.Series): Time values in MM:SS.f format.

    Returns:
        pd.Series: Time values converted to seconds.
    """
    parts = time_series.str.split(":")

    minutes = parts.str[0].astype(float)
    seconds = parts.str[1].astype(float)

    return minutes * 60 + seconds
```

```python
df["start_time_sec"] = time_to_seconds(df["start_time"])
df["end_time_sec"] = time_to_seconds(df["end_time"])
```

```python
df["calculated_end_sec"] = (
    df["start_time_sec"] + df["duration_sec"]
)
```

```python
df["calculated_end_sec"] = (
    df["calculated_end_sec"] % 3600
)
```

```python
df["time_difference"] = (
    df["calculated_end_sec"] - df["end_time_sec"]
)

df["time_difference"].describe().T
```

Therefore, the relationship between `start_time`, `end_time`, and `duration_sec` is consistent, and no changes are required for these columns at this stage.

# <a id="15">  15 ) What is the  Correlation between data_stat variables in the dataset?</a>


```python

"""This time I will use heatmap  to explain how Correlation between data_stat variables

in the dataset in more clear and specific percentage"""

plt.figure(figsize=(10, 8))

cols_for_corr = ['duration_sec', 'member_birth_year', 'member_age', 'bike_id']

df['bike_id'] = df['bike_id'].astype(float) 

corr_matrix = df[cols_for_corr].corr()

sns.heatmap(corr_matrix, annot=True, fmt=".3f", cmap='vlag', vmin=-1, vmax=1, 
            linewidths=1, square=True, cbar_kws={"shrink": .8})

plt.title('Correlation Matrix of Key Numeric Variables', fontsize=16, pad=20)
plt.show()
```

```python
num_cols = ['duration_sec', 'member_age', 'start_station_latitude', 
            'start_station_longitude', 'start_time_sec']

df_multivariate = df[num_cols + ['user_type', 'member_gender']]
```
