```python
def detailed_stats(df, columns):
    stats = []
    for col in columns:
        s = df[col].dropna()
        q1 = s.quantile(0.25)
        q3 = s.quantile(0.75)
        iqr = q3 - q1
        stats.append({
            'Feature': col,
            'Count': int(s.count()),
            'Mean': round(float(s.mean()), 2),
            'Std': round(float(s.std()), 2),
            'Min': round(float(s.min()), 2),
            'Q1 (25%)': round(float(q1), 2),
            'Median (50%)': round(float(s.median()), 2),
            'Q3 (75%)': round(float(q3), 2),
            'Max': round(float(s.max()), 2),
            'IQR': round(float(iqr), 2),
            'Skewness': round(float(s.skew()), 2)
        })
    return pd.DataFrame(stats)

detailed_stats(df_featured, ['duration_min', 'member_age'])
```

# <a id="1"> 1 ) . What is the distribution for User type?</a>

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

# رسم الأعمدة البيانية لنوع المستخدم باستخدام df_final مباشرة
palette = ['#1f77b4', '#aec7e8']
order = df_final['user_type'].value_counts().index
ax = sns.countplot(data=df_final, x='user_type', palette=palette, order=order)

plt.title('Distribution of User Types', fontsize=14, fontweight='bold')
plt.xlabel('User Type', fontsize=12)
plt.ylabel('Number of Trips', fontsize=12)

# حساب الأعداد والنسب المئوية وإضافتها داخل الأعمدة
total_trips = len(df_final)
for p in ax.patches:
  height = p.get_height()
  pct = (height / total_trips) * 100
  ax.annotate(
      f'{int(height):,}\n({pct:.1f}%)',
      (p.get_x() + p.get_width() / 2.0, height / 2),
      ha='center',
      va='center',
      fontsize=11,
      fontweight='bold',
      color='white' if height > 50000 else 'black',
  )

sns.despine(top=True, right=True)
plt.tight_layout()
plt.show()
```

> ### <mark>Your analsys and exploration here </mark>
Subscribers Dominate: Registered Subscribers account for approximately 89.2% of all trips, while casual Customers represent only 10.8%.

Daily Commuter Bias: The service primarily caters to routine daily travel (work or university commutes) rather than tourist or leisure trips.

Conversion Opportunity: Marketing campaigns should focus on converting the 10.8% casual customer pool into recurring subscribers via intro discounts or trial passes.

# <a id="3"> 3 ) What is the distribution for users' ages? </a>

> ### <mark>Your analsys and exploration here </mark>
Right-Skewed Distribution: The age distribution is positively skewed (skewed to the right), with the central peak concentrated between 25 and 35 years old.

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 5))

sns.histplot(data=df_final, x='member_age', bins=30, kde=True, color='#2b5c8f')

mean_age = df_final['member_age'].mean()
median_age = df_final['member_age'].median()

plt.axvline(
    mean_age,
    color='red',
    linestyle='--',
    linewidth=1.5,
    label=f'Mean: {mean_age:.1f}',
)
plt.axvline(
    median_age,
    color='green',
    linestyle='-',
    linewidth=1.5,
    label=f'Median: {median_age:.1f}',
)

plt.title("Distribution of Users' Ages", fontsize=14, fontweight='bold')
plt.xlabel('Member Age (Years)', fontsize=12)
plt.ylabel('Trip Frequency', fontsize=12)
plt.legend(frameon=True)

sns.despine(top=True, right=True)
plt.tight_layout()
plt.show()
```

# <a id="5">  5 ) . What is the distribution for the trip duration in minutes? </a>

```python
## Review the statistic for trip duration in minutes

```

```python
# Make visualization to easy get insight
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 5))

sns.histplot(
    data=df_final,
    x='duration_min',
    bins=60,
    binrange=(0, 60),
    color='#2b5c8f',
    kde=True,
)

mean_dur = df_final['duration_min'].mean()
median_dur = df_final['duration_min'].median()

plt.axvline(
    mean_dur,
    color='red',
    linestyle='--',
    linewidth=1.5,
    label=f'Mean: {mean_dur:.1f} min',
)
plt.axvline(
    median_dur,
    color='green',
    linestyle='-',
    linewidth=1.5,
    label=f'Median: {median_dur:.1f} min',
)

plt.title('Distribution of Trip Duration in Minutes', fontsize=14, fontweight='bold')
plt.xlabel('Trip Duration (Minutes)', fontsize=12)
plt.ylabel('Number of Trips', fontsize=12)
plt.legend(frameon=True)

sns.despine(top=True, right=True)
plt.tight_layout()
plt.show()
```

> ### <mark>Your analsys and exploration here </mark>
Strong Right-Skewed Profile: The distribution of trip durations is heavily skewed to the right, with the vast majority of trips lasting between 5 and 15 minutes and peaking sharply around 8 to 10 minutes.

Median vs. Mean: The median duration is approximately 9.0 minutes, while the mean is slightly higher at roughly 11.5–12.0 minutes, pulled rightward by occasional longer rides


# <a id="6"> 6 ) . What is the distribution for member_gender?</a>

```python
# Make visualization to easy get insight
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

palette = {'Male': '#2b5c8f', 'Female': '#e78ac3', 'Other': '#fc8d62'}

order = df_final['member_gender'].value_counts().index

ax = sns.countplot(
    data=df_final,
    x='member_gender',
    order=order,
    palette=palette
)

plt.title('Distribution of Member Genders', fontsize=14, fontweight='bold')
plt.xlabel('Member Gender', fontsize=12)
plt.ylabel('Number of Trips', fontsize=12)

total_trips = len(df_final)
for p in ax.patches:
    height = p.get_height()
    pct = (height / total_trips) * 100
    
    y_pos = height / 2 if height > 20000 else height + (total_trips * 0.015)
    text_color = 'white' if height > 20000 else 'black'
    
    ax.annotate(
        f'{int(height):,}\n({pct:.1f}%)',
        (p.get_x() + p.get_width() / 2.0, y_pos),
        ha='center',
        va='center',
        fontsize=10,
        fontweight='bold',
        color=text_color
    )

plt.ylim(0, df_final['member_gender'].value_counts().max() * 1.1)

sns.despine(top=True, right=True)
plt.tight_layout()
plt.show()
```

> ### <mark>Your analsys and exploration here </mark>
Male Dominance: Male riders form the largest user segment by a wide margin, accounting for roughly 74.6% of all completed trips.

Female Representation: Female riders represent 23.3% of trips, reflecting a common gender disparity in urban bike-sharing systems.

"Other" Category: Approximately 2.1% (~3,600+ trips) are taken by users who identified as "Other" (non-binary or gender-expansive individuals). While numerically small, it demonstrates platform inclusivity in the diverse San Francisco Bay Area.
